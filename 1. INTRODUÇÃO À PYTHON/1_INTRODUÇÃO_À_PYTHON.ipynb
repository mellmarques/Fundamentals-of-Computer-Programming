3. Fazer uma função que calcula o produto escalar de dois vetores
randômicos a e b de $R^3$, calcula a sua magnitude e determina o ângulo θ que eles formam, usando a fórmula:
$$a · b = ||a|| ||b|| cos θ$$

```python
def prod_escalar():
  import numpy as np
  a = np.random.rand(3)
  b = np.random.rand(3)
  prod = np.dot(a,b)
  print(f"O produtor escalar entre {a} e {b} é {prod:.2f}.")
  mod_a = np.linalg.norm(a)
  mod_b = np.linalg.norm(b)
  ang1 = np.arccos(prod/(mod_a*mod_b))
  ang2 = np.degrees(ang1)
  print(f"O ângulo entre os vetores é igual a {ang2:.2f}°.")

prod_escalar()
```

  - Pega uma matriz randômica A de n × n e calcula a sua m-essima potência (tomar valores de m = 2, 3, 4).

- Plotar o tempo de

- No segundo ponto, colocar no mesmo
gráfico os resultados para os diferentes valores de m (tomar valores de $m=500, 1000, 1500$).

```python
import numpy as np
import matplotlib.pyplot as plt
import time

# PARTE 1: OPERAÇÃO COM VETORES

dimensoes_vetores = [10**4, 10**5, 10**6, 10**7] # dimensões dadas de exemplo, eixo x

tempos_vetor = [] # vai ser o eixo y
for n in dimensoes_vetores:
  a = np.random.rand(n) # cria array com n elementos de 0 a 1
  b = np.random.rand(n)
  alpha = np.random.rand() # cria um elemento aleatorio entre 0 e 1
  beta = np.random.rand()
  # executando a operação
  inicio = time.time() # salva o tempo que foi acionado
  c = alpha * a + beta * b
  fim = time.time()
  # adiciona o tempo desse n na lista
  tempo_decorrido = fim - inicio
  tempos_vetor.append(tempo_decorrido)

# criar uma figura com 2 subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Gráfico 1: Escala Linear
# plotar as duas listas de mesmo tamanho, marcando esses valores com um marker, e ligando eles
# o .plot liga as coordenadas dadas com uma linha
ax1.plot(dimensoes_vetores, tempos_vetor, marker='o')
ax1.set_title('Dimensão X Tempo de operação (Escala Linear)')
ax1.set_xlabel('Dimensão n')
ax1.set_ylabel('Tempo (s)')
ax1.grid(True)

# Gráfico 2: Escala Log-Log
ax2.plot(dimensoes_vetores, tempos_vetor, marker='o')
ax2.set_xscale('log') # mudando a escala do loglog
ax2.set_yscale('log')
ax2.set_title('Dimensão X Tempo de operação (Escala Log-Log)')
ax2.set_xlabel('Dimensão n')
ax2.set_ylabel('Tempo (s)')
ax2.grid(True, which="both") # escala log precisa de grade secundaria tbm

plt.show()

# PARTE 2: OPERAÇÃO COM MATRIZES

dimensoes_matriz = [10, 200, 500, 800]
potencias = [500, 1000, 1500]

# criando um dicionário que armazena os tempos de cada potencia m
tempos_matriz = {m: [] for m in potencias}
for m in potencias:
  for n in dimensoes_matriz:
    A = np.random.rand(n, n) # cria uma matriz quadrada de dimensão n, com valores de 0 a 1

    inicio = time.time()
    B = np.linalg.matrix_power(A, m) # função numpy de elevar uma matriz a uma potencia
    fim = time.time()

    tempo_decorrido = fim - inicio
    tempos_matriz[m].append(tempo_decorrido) # adiciona o valor achado à sua chave m correspondente

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

# Gráfico 1: Escala Linear
# para pegar so os valores do dicionario, percorre todas as chaves
for m in potencias:
  ax1.plot(dimensoes_matriz, tempos_matriz[m], marker='o', label=f'm = {m}')
  # o label vai definir a configuração da legenda dessa linha que acabei de plotar
ax1.set_title('Tempo de Cálculo vs. Dimensão da Matriz (Escala Linear)')
ax1.set_xlabel('Dimensão n')
ax1.set_ylabel('Tempo (s)')
ax1.legend() # ativa a legenda que eu configurei no plot
ax1.grid(True)

# Gráfico 2: Escala Log-Log
for m in potencias:
  ax2.plot(dimensoes_matriz, tempos_matriz[m], marker='o', label=f'm = {m}')
ax2.set_xscale('log')
ax2.set_yscale('log')
ax2.set_title('Tempo de Cálculo vs. Dimensão da Matriz (Escala Log-Log)')
ax2.set_xlabel('Dimensão n')
ax2.set_ylabel('Tempo (s)')
ax2.legend()
ax2.grid(True, which="both")

plt.tight_layout() # isso aqui deixa o grafico bonitinho sem sobrepor

plt.show()

```

###Q2

- Fazer uma função que recebe uma matriz randômica A de
dimensão m × r e outra matriz randômica B de dimensão r × n,
verifica as suas dimensões e realiza a multiplicação delas no
sentido usual de algebra linear, para retornar uma matriz C de
dimensão m × n. A multiplicação deve ser feita usando todos
os for que sejam necessários, lembrando que a fórmula para
calcular o coeficiente $c_{ij}$ é dado por

$$c_{ij}=\sum_{k=1}^{r} a_{ik}b_{kj}, (i = 1, ..., m), (j = 1, ..., n)$$  

- Considerando matrizes quadradas (m = r = n), medir o
tempo de cálculo como função da dimensão e comparar com o
tempo necessário fazendo A @ B (tomar dimensões de matriz 50, 100, 150, 200, 250). E plotar o tempo de cálculo
como função da dimensão na escala loglog.


```python
import numpy as np
import time
import matplotlib.pyplot as plt

# PARTE 1 ----------------------------------------------------------------------------------

# A = m x r (m linhas, r colunas)
# B = r x n (r linhas, n colunas)

# criando uma matriz A (2x3) com números aleatórios inteiros de 0 a 9
# o size=(x,y) define o numero de linhas e colunas
matriz_A = np.random.randint(0, 10, size=(2, 3))

# criando uma matriz B (3x4) com números aleatórios inteiros de 0 a 9
matriz_B = np.random.randint(0, 10, size=(3, 4))

def multiplica_matrizes(A, B):
  # verificar r = r
  num_linhas_B = len(B) # conta quantos vetores tem em B
  num_colunas_A = len(A[0]) # conta quantos itens tem no primeiro vetor (primeiro = 0) de A, que igual a qualquer vetor

  if num_colunas_A != num_linhas_B:
    print('Dimensões incompatíveis')
    return None # trava o programa

  r = len(B) # = len(A[0]) que ja verificamos
  m = len(A) # linhas (quantos vetores)
  n = len(B[0]) # colunas (quantos itens por vetor)

  # Criar uma matriz C de zeros (espaços vazios) do tamanho mxn
  C = np.zeros((m, n))

  # i percorre as LINHAS de C (0, 1, 2, ..., m)
  for i in range(m):
    # j percorre as COLUNAS de C (0, 1, 2, ..., n)
    for j in range(n):
      # usar a formula para calcular cada elemento C[i, j], e depois substituir do lugar do zero que tava
      soma = 0
      for k in range(r):
        soma += A[i, k] * B[k, j] # produto de matrizes
        C[i, j] = soma
  return C.astype(int) # converter toda essa matriz float para números inteiros

print('Matriz A:')
print(matriz_A)
print('Matriz B:')
print(matriz_B)

print('Matriz C:')
# usando a função
matriz_C = multiplica_matrizes(matriz_A, matriz_B)
print(matriz_C)

# PARTE 2 ---------------------------------------------------------------------------------

# medir o tempo para cada uma das dimensões
dimensoes = [50, 100, 150, 200, 250] # eixo x

tempos_manual = [] # eixo y1
tempos_numpy = [] # eixo y2

for n in dimensoes:
  matriz_A = np.random.randint(0, 10, size=(n, n))
  matriz_B = np.random.randint(0, 10, size=(n, n))
  # tempo manual
  inicio = time.time()
  # usa a função
  resultado = multiplica_matrizes(matriz_A, matriz_B)
  fim = time.time()
  tempo = fim - inicio
  tempos_manual.append(tempo)
  # tempo numpy
  inicio = time.time()
  resultado = matriz_A @ matriz_B
  fim = time.time()
  tempo = fim - inicio
  tempos_numpy.append(tempo)

# criando a figura com 2 subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# gráfico linear
ax1.plot(dimensoes, tempos_manual, marker='o', label='Função Manual') # o bolinha, s quadrado
ax1.plot(dimensoes, tempos_numpy, marker='s', label='Função NumPy')
ax1.set_title('Tempo de Cálculo x Dimensão (Escala Linear)')
ax1.set_xlabel('Dimensão da Matriz (n)')
ax1.set_ylabel('Tempo (s)')
ax1.grid(True)
ax1.legend()

# gráfico loglog
ax2.plot(dimensoes, tempos_manual, marker='o', label='Função Manual')
ax2.plot(dimensoes, tempos_numpy, marker='s', label='Função NumPy')
# mudar as escalas!
ax2.set_xscale('log')
ax2.set_yscale('log')

ax2.set_title('Tempo de Cálculo x Dimensão (Escala Log-Log)')
ax2.set_xlabel('Dimensão da Matriz (n)')
ax2.set_ylabel('Tempo (s)')
ax2.grid(True, which='both')
ax2.legend()

plt.tight_layout()
plt.show()
```

###Q3
- Fazer um gráfico que mostra pontos distribuídos randomicamente dentro da região definida por x = 0, y = 0 e y = 1 − x. Mostrar diferentes casos, considerando diferentes quantidades
de pontos.

```python
import numpy as np
import matplotlib.pyplot as plt

# triangulo de verticies (0,0), (1,0), (0,1)
# criar uma função que gere a quantidade de pontos desejados dentro do triangulo
def gerar_pontos(num):
  # criar duas listas que serao os eixos do grafico
  pontos_x = []
  pontos_y = []
  pontos_gerados = 0
  while pontos_gerados < num:
    # a estrategia é pegar um ponto aleatório dentro do quadrado [0,1] x [0,1]
    x = np.random.rand()
    y = np.random.rand()
    # e depois verificar se esse ponto está dentro do triângulo (y+x=1)
    if (x + y) <= 1:
      pontos_x.append(x)
      pontos_y.append(y)
      # se tiver, contamos este ponto como válido
      pontos_gerados += 1
  # isso se repete ate ter a quantidade de pontos que pedi no começo
  # transformar as listas em array para usar depois a função do matplotlib (scatter)
  pontos_x = np.array(pontos_x)
  pontos_y = np.array(pontos_y)

  return pontos_x, pontos_y

# queremos um grafico para cada uma dessas quantidades
lista_quantidades = [50, 200, 1000, 5000]

# vamos criar uma figura com 4 subplots, um para cada valor da lista
fig, axs = plt.subplots(2, 2, figsize=(12, 12))

# flat() percorre a matriz dos 4 eixos axs = [ [area_00, area_01], [area_10, area_11] ]
# na ordem [area_00, area_01, area_10, area_11] de indices 0, 1, 2, 3
axs = axs.flat

indice = 0
for i in lista_quantidades:
  # definindo qual grafico vamos plotar
  ax = axs[indice]
  # usando a função
  x_pontos, y_pontos = gerar_pontos(i)
  indice += 1
  # a funçao scatter serve para desenhar pontos aleatórios no gráfico
  ax.scatter(x_pontos, y_pontos, s=4, alpha = 0.6, color='red') # s=tamanho, alpha=transparencia

  # desenhar as retas que definem o triangulo para cada grafico
  # ax.plot( [x_inicio, x_fim], [y_inicio, y_fim] )
  ax.plot([0, 1], [0, 0], color='black')  # eixo x
  ax.plot([0, 0], [0, 1], color='black')  # eixo y
  ax.plot([0, 1], [1, 0], color='black')  # y=1-x

  ax.set_title(f'{i} Pontos no Triângulo')
  ax.set_xlabel('X')
  ax.set_ylabel('Y')
  ax.grid(True, linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def calcula_sequencia(num):
  soma = 0
  for n in range(num):
    termo = (-1)**n/2*n + 1
    soma += termo
  pi_calculado = soma * 4
  return pi_calculado

valores_de_N = [10, 100, 1000, 10000, 100000, 1000000]
lista_de_erros = []

for N in valores_de_N:
  pi_calculado = calcula_sequencia(N)
  erro = abs(np.pi - pi_calculado)
  lista_de_erros.append(erro)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# grafico linear
ax1.plot(valores_de_N, lista_de_erros, marker='o')
ax1.set_title('Erro no cálculo de Pi (Escala Linear)')
ax1.set_xlabel('Número de Termos (N)')
ax1.set_ylabel('Erro Absoluto')
ax1.grid(True)

# grafico loglog
ax2.plot(valores_de_N, lista_de_erros, marker='o')
ax2.set_xscale('log')
ax2.set_yscale('log')
ax2.set_title('Erro no cálculo de Pi (Escala Log-Log)')
ax2.set_xlabel('Número de Termos (N)')
ax2.set_ylabel('Erro Absoluto')
ax2.grid(True, which='both')

plt.tight_layout()
plt.show()
```

- Considerar uma sequência de numeros gerada
da seguinte forma:
$$x_n=a x_{n-i}(1-x_{n-1}), (n = 1, 2, . . . , 5000), x_0 = 0.1$$


- Considerar diferentes valores de "a" entre
0 e 4 (p.e., a = 1, 2, 3.2, 3.5 e 4). Calcular a média e a variança da
sequência:

$$\bar{x}=\frac{1}{N}\sum_{i=0}^{N} x_i$$
$$σ =\frac{1}{N-1}\sum_{i=0}^{N} (x_i-\bar{x})^2$$

- Programar-lo na mão e usando funções de numpy já prontas (np.mean e np.var) e comparar os resultados.



```python
import numpy as np

N = 5000
x0 = 0.1
valores_de_a = [1, 2, 3.2, 3.5, 4]

for a in valores_de_a:
  # adicionar toda a sequencia numa lista
  sequencia = [x0]
  x_anterior = x0
  for n in range(N):
    x_novo = a * x_anterior * (1 - x_anterior)
    sequencia.append(x_novo)
    x_anterior = x_novo

  # calculando a média manual:
  soma = 0
  for x in sequencia:
    soma += x
  media_manual = soma / N

  # calculando a variancia manual:
  soma = 0
  for x in sequencia:
    soma += (x - media_manual)**2
  variancia_manual = soma / (N - 1)

  # usando funções numpy:
  sequencia_np = np.array(sequencia) # para usar np, transformamos as listas em arrays
  media_numpy = np.mean(sequencia_np)
  variancia_numpy = np.var(sequencia_np, ddof=1) # ddof=1 significa denominador N-1, por padrao eh N-0

  print(f'a = {a}')
  print(f"Média (manual): {media_manual:.8f}")
  print(f"Média (NumPy): {media_numpy:.8f}")
  print(f"Variância (manual): {variancia_manual:.8f}")
  print(f"Variância (NumPy): {variancia_numpy:.8f}")
  print('---------------------------------')

```

$$0.1 0.0$$

- Para carregar o arquivo deverá ser usado o comando
meuspontos = np.loadtxt(’meuarquivo.txt’)

```python
import numpy as np
import matplotlib.pyplot as plt

def plotar_poligono_fechado(nome_do_arquivo):
    pontos = np.loadtxt(nome_do_arquivo) # transforma esses números do arquivo em uma "matriz" (array).
    # : significa "todas as linhas"
    coordenadas_x = pontos[:, 0] # uma lista com os primeiros elementos de todas as linhas
    coordenadas_y = pontos[:, 1] # agora pra segunda coluna

    # o tamanho vai depender do arquivo
    fig, ax = plt.subplots()

    # adicionar ao final da lista a primeira coordenada dela
    # pois assim podemos ligar a primeira na ultima e fechar o poligono
    x_linha = np.append(coordenadas_x, coordenadas_x[0])
    y_linha = np.append(coordenadas_y, coordenadas_y[0])
    # desenha as linhas que ligam os pontos do poligono em azul
    ax.plot(x_linha, y_linha, linestyle='-', color='blue', label='Região Poligonal')
    # desenhar bolinhas em vermelho nos pontos
    ax.plot(x_linha, y_linha, marker='o', linestyle='', color='red', label='Pontos do Arquivo')

    ax.set_title("Desenho da Região Poligonal")
    ax.set_xlabel("Eixo X")
    ax.set_ylabel("Eixo Y")
    ax.grid(True)
    ax.legend()
    ax.set_aspect('equal', adjustable='box') # para as escalar nao distorcerem
    plt.show()

# usando a função
arquivo_com_pontos = 'meuarquivo.txt'
plotar_poligono_fechado(arquivo_com_pontos)
```

###Q10
- Considerar o lançamento de uma bala de canhão. Fazer um programa
que recebe o ângulo e velocidade de lançamento da bala e plotar
a trajetória da bala. O código deveria funcionar para poder
plotar no mesmo gráfico a trajetória correspondente a vários
valores dos parâmetros de entrada.

```python
import numpy as np
import matplotlib.pyplot as plt

vo = float(input('Velocidade de lançamento (em m/s): '))
ang = float(input('ângulo de lançamento (em graus): '))
g = 9.81

# fórmulas
ang_rad = np.radians(ang)
vo_x = vo*np.cos(ang_rad)
vo_y = vo*np.sin(ang_rad)
T = abs(2*vo_y/g)
alcance = vo_x*T
h = vo_y*(T/2)-0.5*g*((T/2)**2)

print(f'Altura máxima: {h:.2f} m')
print(f'Alcance: {alcance:.2f} m')
print(f'Tempo total no ar: {T:.2f} s')

def calcular_trajetoria(vo, ang):
  # pegar 200 instantes entre 0 e tempo_total:
  t = np.linspace(0, T, 100)
  x = vo_x*t
  y = vo_y*t-0.5*g*(t**2)
  return x, y

# chamando a função nos valores dados
x, y = calcular_trajetoria(vo, ang)

# criando o grafico
fig, ax = plt.subplots()
ax.plot(x, y, label=f'v = {vo} m/s, Ângulo = {ang}°')
ax.set_title('Trajetória de uma Bala de Canhão para Diferentes Lançamentos')
ax.set_xlabel('Distância Horizontal (m)')
ax.set_ylabel('Altura Vertical (m)')
ax.grid(True, which='both')
ax.legend()

plt.show()
```

- Desenhar várias trajetórias de partículas que realizam 1000 passos. Para escolher a direção usar a função
np.random.randint(4).

```python
import numpy as np
import matplotlib.pyplot as plt

passos = 1000
distancia = 1
particulas = 5

# todas as particulas vao ficar no mesmo ax
fig, ax = plt.subplots(figsize=(8, 8))

for i in range(particulas):
  # começar no 0,0
  posicao_x = 0
  posicao_y = 0
  # criar a lista do eixo x e eixo y
  caminho_x = [posicao_x]
  caminho_y = [posicao_y]
  for i in range(passos):
    direcao = np.random.randint(4) # temos 4 opçoes = (0,1,2,3)
    if direcao == 0:
      posicao_y += distancia # cima
    elif direcao == 1:
      posicao_y -= distancia # baixo
    elif direcao == 2:
      posicao_x += distancia # direita
    else:
      posicao_x -= distancia # esquerda
    caminho_x.append(posicao_x)
    caminho_y.append(posicao_y)
  # plotar as duas listas dessa particula quando terminar os passos
  ax.plot(caminho_x, caminho_y, label=f'Partícula {i+1}')

# configurar o gráfico final
ax.set_title(f'Trajetórias aleatória de {particulas} Partículas')
ax.set_xlabel('Posição X')
ax.set_ylabel('Posição Y')
ax.grid(True)
# Adiciona um ponto vermelho grande na origem para marcar o início
ax.plot(0, 0, color='red', marker='o', markersize=10, label='Início (0,0)')
ax.legend()
# deixa sempre a mesma escala em x e y pra nao distorcer
ax.set_aspect('equal', adjustable='box')

plt.show()
```

um grafo. Notar que em geral ela estará caracterizada por um
certo número de nós (ou uniões), um certo nuúmero de arestas e

- Fazer uma estrutura de dados que sirva para descrever essa

matriz ou array que indica como os nós e as arestas estão
relacionados. Adicionalmente, a estrutura deve conter um

Construir um exemplo inventando as coordenadas e plotar

uma aresta da rede. Notar que se a aresta possui um nó
que náo pertence a nenhuma outra aresta, esse nó também

```python
import matplotlib.pyplot as plt

'''
usar um dicionario principal (rede), que contenha dois dicionário (coordenadas) e (conexoes)
em coordenada a chave é o nó e o valor é sua posição (x,y)
em conexões a chave é o no e o valor é uma lista com todos os nós que ele se conecta
'''

rede = {'coordenadas': {0: (0, 2), 1: (1, 2.5), 2: (2.5, 2.5), 3: (1, 1.5), 4: (2.5, 1.5), 5: (3.5, 1), 6: (4, 3), 7: (5, 2), 8: (6, 3), 9: (6, 1)},
        'conexoes': {0: [1], 1: [0, 2, 3], 2: [1, 4, 6], 3: [1, 4], 4: [2, 3], 5: [6, 7], 6: [2, 5, 7], 7: [5, 6, 8], 8: [7, 9], 9: [8]}}

# PARTE 1 ------------------------------------------------------------------------

def plotar_rede(dados_rede):
  # separa o dicionario em dois
  coordenadas = dados_rede['coordenadas'] # dic1
  conexoes = dados_rede['conexoes'] #dic2
  # cria a figura
  fig, ax = plt.subplots(figsize=(12, 7))
  # abrindo o dicionario 'conexoes'
  # chamando as chaves de nó e as listas de 'vizinhos'

  for no, vizinhos in conexoes.items():
    # ex: no=2 e vizinhos=[1, 4, 6]
    x1, y1 = coordenadas[no] # achar as coordenadas desse nó
    # ex: no=2, x1,y1=(2.5, 2.5)
    for i in vizinhos:
      # ex: no=2, i = 1, 4, 6
      # condição para evitar desenhar a mesma linha duas vezes
      if no < i:
      # 2<1 --> pula, 2<4 --> confere, 2<6 confere
      # pula o vizinho 1 pois eu ja desenhei a linha que conecta eles antes
        x2, y2 = coordenadas[i]
        # ex: coordenada[4] = (2.5, 1.5)
        # desenha as linhas que ligam os nós
        ax.plot([x1, x2], [y1, y2], color='gray')

  # desenha as bolinhas dos nós
  for no, coord in coordenadas.items():
    # chama a chave de no e as coordenadas de coord (tupla)
    x, y = coord # abre a tupla
    ax.plot(x, y, marker='o', color='pink', markersize=20, markeredgecolor='black')
    # escreve o numero do nó dentro da coordenada x, y
    # alinha no centro horizontalmente (ha) e verticalmente(va), e deixa em negrito
    ax.text(x, y, str(no), color='black', ha='center', va='center', fontweight='bold')

  # Configurações do gráfico
  ax.set_title('Visualização da Rede')
  ax.set_xlabel('Coordenada X')
  ax.set_ylabel('Coordenada Y')
  ax.set_aspect('equal', adjustable='box')
  plt.grid(True)
  plt.show()

# PARTE 2 ------------------------------------------------------------------------

def adicionar_aresta(dados_rede, no1, no2):
  conexoes = dados_rede['conexoes']
  if no1 not in conexoes or no2 not in conexoes:
      print(f"Erro: Um dos nós ({no1} ou {no2}) não existe na rede.")
      return
  if no2 in conexoes[no1]:
      print(f"A aresta entre {no1} e {no2} já existe.")
      return
  conexoes[no1].append(no2)
  conexoes[no2].append(no1)

# PARTE 3 ------------------------------------------------------------------------

def apagar_aresta(dados_rede, no1, no2):
  conexoes = dados_rede['conexoes']
  if no1 not in conexoes or no2 not in conexoes:
      print(f"Erro: Um dos nós ({no1} ou {no2}) não existe na rede.")
      return
  if no2 not in conexoes[no1]:
      print(f"Erro: A aresta entre {no1} e {no2} não existe.")
      return
  conexoes[no1].remove(no2)
  conexoes[no2].remove(no1)

# EXECUÇÃO DO PROGRAMA -------------------------------------------------------------

# PARTE 1
plotar_rede(rede)
# PARTE 2
adicionar_aresta(rede, 4, 5)
plotar_rede(rede)
# PARTE 3
apagar_aresta(rede, 6, 7)
plotar_rede(rede)

```
